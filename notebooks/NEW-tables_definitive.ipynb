# WaveletFlow — Results Tables

Reproduces the paper's results tables from the per-experiment metric CSVs, in
two flavours from the **same** builders:

* **paper** — one run, all models, matches the numbers in the paper. Reads `*_stats.csv`.
* **seeds** — mean ± std across seeds, only the best model per PDE (for the rebuttal). Reads `*_metrics.csv`.

All supporting code lives in **`table_utils.py`** (imported once below), so this
notebook stays a thin list of table definitions. Paper-table map:

| Section | Paper table |
|---|---|
| VRMSE by horizon | S11 |
| CRPS by horizon | S12 |
| Spectral RMSE by band | S13 (S9) |
| Wavelet selection | Table 1 / S10 |
| CRPS ablations | Table 3a/S14, 3b/S16, S17 |
| Spectral ablations | S15, S18 |

Timing tables (2, S19, S20) come from separate profiling and are not built here.

## 1 · Load the data

Missing models are reported and skipped — expected while some multi-seed runs are still landing.

```python
stats_df   = T.load(T.SINGLE_SUFFIX) if EMIT_PAPER else None
metrics_df = T.load(T.SEEDS_SUFFIX, models=ALL_ROWS + [LATENTFM]) if EMIT_SEEDS else None

# Thin wrapper so each table below is a single call.
def table(name, build_fn, latex_fn, caption, label, **kw):
    kw.setdefault("run_paper", EMIT_PAPER)
    kw.setdefault("run_seeds", EMIT_SEEDS)
    return emit(name, build_fn, latex_fn, caption, label,
                paper_df=stats_df, seeds_df=metrics_df, **kw)
```

## 5 · Wavelet selection — Table 1 / S10

Band-averaged spectral coherence RMSE per PDE. **paper** (needs the single-run
stats, all four wavelets): **bold** marks the best wavelet *within each WFM
scale* — the pick reported in Table S10 — driven together with `BEST_WFM_PER_PDE`
by one `select_wavelets()`, so the three always agree. **seeds** gives the same
band-averaged quantity for the best model per PDE with mean ± std; the wavelet
*selection* itself isn't re-done across seeds (those runs only contain the
already-selected wavelet).

Note: the per-PDE value is the mean of the *unrounded* Low/Mid/High bands, so it
can differ by a thousandth from averaging the three rounded numbers in Table S13
— this matches the paper (e.g. 0.119 vs 0.120).

```python
emit_selection("T1_selection", stats_df, models=ALL_ROWS_GEN, groups=GROUPS_ALL_GEN,
               caption=r"Spectral Coherence RMSE $\downarrow$ averaged across Low, Mid "
                       r"and High frequency bands per PDE system. Bold: optimal wavelet "
                       r"per WFM variant.",
               label="tab:selection")

emit_selection_seeds("T1_selection", metrics_df,
                     caption=r"Spectral Coherence RMSE $\downarrow$ averaged across Low, "
                             r"Mid and High frequency bands per PDE system, best model per PDE.",
                     label="tab:selection")
```

### 8b · Seed-spread check

If the across-seed std comes out ~0, use this to see whether the seeds actually
differ. It prints the per-seed value of one cell; identical values mean the
metric files barely vary by seed (a data property, not the aggregation).

```python
if metrics_df is not None:
    spread = seed_spread(build_horizon_table, metrics_df, "crps", models=BEST_ROWS_GEN)
    cell = ("rayleigh_benard", HORIZONS["rayleigh_benard"][0])
    print("Per-seed CRPS, WFM 1-scale, RB", cell[1], ":")
    for s, tbl in spread.items():
        v = tbl.loc[WFM1_BEST, cell] if WFM1_BEST in tbl.index else float("nan")
        print(f"  seed {s}: {v:.6f}")
```
