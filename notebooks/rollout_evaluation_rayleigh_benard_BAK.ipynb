```python
well_dataset_name = 'rayleigh_benard'

model_checklist = [
    'WaveletFlow_1scale_db2_v4.0',
    'WaveletFlow_3scales_db2_v4.0',
    'FM_data_space_v4.0',
    #'FNO',
    #'TFNO',
    #'UNetClassic',
    'UNetConvNext',
    ]

generative_models = [
    'WaveletFlow_1scale_db2_v4.0',
    'WaveletFlow_3scales_db2_v4.0',
    'FM_data_space_v4.0',
    ]
```

```python
GT_LABEL = "Ground-truth"

dfs_stats, dfs_psd = [], []

for model in model_checklist:
    pattern_stats = f'{PROJECT_ROOT}/experiments/{well_dataset_name}/metrics/{model}*stats.csv'
    pattern_psd   = f'{PROJECT_ROOT}/experiments/{well_dataset_name}/metrics/{model}*psd.csv'
    matched_files_stats = glob.glob(pattern_stats)
    matched_files_psd   = glob.glob(pattern_psd)

    if not matched_files_stats:
        print(f"[WARN] No files found for model '{model}' with pattern: {pattern_stats}")
        continue
    if not matched_files_psd:
        print(f"[WARN] No files found for model '{model}' with pattern: {pattern_psd}")
        continue

    def enrich(df, stem, model_name):
        """Add metadata columns, preserving GT rows already labelled in the CSV."""
        parts     = stem.split('_')
        nsampling = next((p.replace('nsampling', '') for p in parts if p.startswith('nsampling')), None)
        nens      = next((p.replace('nens',      '') for p in parts if p.startswith('nens')),      None)

        # Only overwrite model label for rows that are NOT already ground truth
        if 'model' in df.columns:
            df.loc[df['model'] != GT_LABEL, 'model'] = model_name
        else:
            df['model'] = model_name

        df['source']    = stem
        df['dataset']   = well_dataset_name
        df['nsampling'] = int(nsampling) if nsampling else None
        df['nens']      = int(nens)      if nens      else None
        return df

    for fpath in sorted(matched_files_stats):
        try:
            df   = pd.read_csv(fpath)
            stem = Path(fpath).stem
            df   = enrich(df, stem, model)
            dfs_stats.append(df)
            print(f"  [OK] {df.shape} | {Path(fpath).name:70s}")
        except Exception as e:
            print(f"  [ERR] {fpath}: {e}")

    for fpath in sorted(matched_files_psd):
        try:
            df   = pd.read_csv(fpath)
            stem = Path(fpath).stem
            df   = enrich(df, stem, model)
            dfs_psd.append(df)
            print(f"  [OK] {df.shape} | {Path(fpath).name:70s}")
        except Exception as e:
            print(f"  [ERR] {fpath}: {e}")

if dfs_stats:
    stats_df = pd.concat(dfs_stats, ignore_index=True)
    # Ground-truth rows appear once per source file — deduplicate them
    stats_df = stats_df.drop_duplicates(
        subset=[c for c in stats_df.columns if c not in ('source', 'nsampling', 'nens')]
    )
    print(f"\nMerged DataFrame: {stats_df.shape}  |  columns: {list(stats_df.columns)}")
else:
    stats_df = pd.DataFrame()
    print("[WARN] No data loaded — stats_df is empty.")

if dfs_psd:
    psd_df = pd.concat(dfs_psd, ignore_index=True)
    psd_df = psd_df.drop_duplicates(
        subset=[c for c in psd_df.columns if c not in ('source', 'nsampling', 'nens')]
    )
    print(f"\nMerged DataFrame: {psd_df.shape}  |  columns: {list(psd_df.columns)}")
else:
    psd_df = pd.DataFrame()
    print("[WARN] No data loaded — psd_df is empty.")
```

# Plot: VRMSE at different rollout steps/regimes

```python
det_lookup = (
    stats_df[["model", "model_type"]]
    .drop_duplicates()
    .set_index("model")["model_type"]
    .to_dict()
)

horizons   = sorted(stats_df["horizon"].unique())
all_models = stats_df["model"].unique().tolist()
n_models   = len(all_models)
bar_width  = 0.8 / n_models

fig, axes = plt.subplots(
    1, n_fields,
    figsize=(2.4 * n_fields, 2.4 + 0.6),
    squeeze=False,
)
axes_flat = axes.flatten()

for ax, field in zip(axes_flat, field_names):
    table = stats_df[stats_df["field"] == field]

    for h_idx, horizon in enumerate(horizons):
        h_table = table[table["horizon"] == horizon]

        # aggregate + sort for this (field, horizon) group
        model_vals = (
            h_table.groupby("model")["vrmse"]
            .apply(aggregate)
            .sort_values()          # shortest bar first (best model)
            .dropna()
        )

        n_present = len(model_vals)
        offsets   = np.linspace(
            -(n_present - 1) / 2,
             (n_present - 1) / 2,
            n_present,
        ) * bar_width

        for offset, (model_name, val) in zip(offsets, model_vals.items()):
            is_det = det_lookup.get(model_name) == "deterministic"
            ax.bar(
                h_idx + offset, val,
                width=bar_width,
                color=model_palette[model_name],
                hatch="//" if is_det else "",
                edgecolor="white" if is_det else model_palette[model_name],
            )

    ax.set_xticks(range(len(horizons)))
    ax.set_xticklabels(horizons)
    ax.grid(axis="y", linestyle="--", alpha=0.3)
    ax.set_xlabel("Lead time")
    ax.set_ylabel("VRMSE" if ax is axes_flat[0] else "")
    ax.set_title(field)

# ── shared legend sorted by global mean VRMSE ─────────────────────────────
global_order = (
    stats_df.groupby("model")["vrmse"]
    .apply(aggregate)
    .sort_values()
    .index.tolist()
)

legend_handles = [
    mpatches.Patch(
        facecolor=model_palette[m],
        hatch="//" if det_lookup.get(m) == "deterministic" else "",
        edgecolor="white" if det_lookup.get(m) == "deterministic" else model_palette[m],
        label=m,
    )
    for m in global_order
]

fig.legend(
    handles=legend_handles,
    loc="lower center",
    bbox_to_anchor=(0.5, -0.02),
    ncol=4,
    frameon=True,
    fontsize="small",
)

fig.tight_layout(pad=0.33)
fig.subplots_adjust(bottom=0.28)
#plt.savefig(f"{PROJECT_ROOT}/experiments/{well_dataset_name}/figures/vrmse_rollout_all_fields.png", dpi=300, bbox_inches='tight') #transparent=True)
plt.show()
```

```python
"""
Panel D — Absolute VRMSE across physical regimes (Rayleigh × Prandtl)
Layout  : one heatmap per model on a shared colour scale
Metric  : VRMSE averaged over all fields, trajectories, and rollout steps
Reading : darker teal = lower error = better
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from mpl_toolkits.axes_grid1 import make_axes_locatable

# ── 0. Config ─────────────────────────────────────────────────────────────────

HERO = "WaveletFlow_3scales_db2_v4.0"
MODELS = [
    "WaveletFlow_3scales_db2_v4.0",
    "WaveletFlow_1scale_db2_v4.0",
    "FM_data_space_v4.0",
    "UNetConvNext",
]
MODEL_LABELS = [
    "WaveletFlow\n3-scales (ours)",
    "WaveletFlow\n1-scale",
    "FM\ndata-space",
    "UNetConvNext",
]

TEAL_DARK  = np.array([  4/255, 52/255,  44/255, 1.0])   # #04342C  (low error)
TEAL_MID   = np.array([ 29/255,158/255, 117/255, 1.0])   # #1D9E75
TEAL_LIGHT = np.array([225/255,245/255, 238/255, 1.0])   # #E1F5EE  (high error)

# ── 1. Decode log-stored Ra / Pr ──────────────────────────────────────────────

reg_df = stats_df[stats_df["model"].isin(MODELS)].copy()

def round_to_order(s):
    """Collapse float noise: 9.9999e8 → 1e9."""
    return 10.0 ** np.log10(s.clip(lower=1e-30)).round()

reg_df["Ra"] = round_to_order(np.exp(reg_df["Rayleigh"]))
reg_df["Pr"] = round_to_order(np.exp(reg_df["Prandtl"]))

# ── 2. Aggregate: mean VRMSE per (model, Ra, Pr) ─────────────────────────────

agg = (
    reg_df.groupby(["model", "Ra", "Pr"])["vrmse"]
    .mean()
    .reset_index()
)

def make_grid(model_name):
    g = (
        agg[agg["model"] == model_name]
        .pivot(index="Pr", columns="Ra", values="vrmse")
        .sort_index(axis=0)
        .sort_index(axis=1)
    )
    return g

grids = [make_grid(m) for m in MODELS]

# Align all grids to the same (Ra, Pr) index
ref = grids[0]
grids = [g.reindex(index=ref.index, columns=ref.columns) for g in grids]

ra_vals = ref.columns.values
pr_vals = ref.index.values
n_pr, n_ra = ref.shape

def sci_label(v):
    exp = int(np.round(np.log10(v)))
    return r"$1$" if exp == 0 else rf"$10^{{{exp}}}$"

ra_labels = [sci_label(v) for v in ra_vals]
pr_labels = [sci_label(v) for v in pr_vals]

# ── 3. Shared colour scale (teal: light = high error, dark = low error) ───────

cmap = mcolors.LinearSegmentedColormap.from_list(
    "teal_vrmse", [TEAL_DARK, TEAL_MID, TEAL_LIGHT], N=256
)

all_vals = np.concatenate([g.values.ravel() for g in grids])
vmin = np.nanmin(all_vals)
vmax = np.nanmax(all_vals)

# ── 4. Figure: 1 row × N_models subplots ────────────────────────────────────

plt.rcParams.update({
    "font.family"     : "DejaVu Sans",
    "font.size"       : 9,
    "axes.linewidth"  : 0.5,
    "xtick.direction" : "out",
    "ytick.direction" : "out",
    "xtick.major.size": 2.5,
    "ytick.major.size": 2.5,
    "figure.dpi"      : 150,
})

n_models = len(MODELS)
fig, axes = plt.subplots(
    1, n_models,
    figsize=(2.8 * n_models, 3.2),
    gridspec_kw={"wspace": 0.08},
)

for col, (ax, grid, label, model) in enumerate(
        zip(axes, grids, MODEL_LABELS, MODELS)):

    im = ax.imshow(
        grid.values,
        cmap=cmap,
        vmin=vmin, vmax=vmax,
        aspect="auto",
        origin="lower",
        interpolation="nearest",
    )

    # Cell annotations
    for r in range(n_pr):
        for c in range(n_ra):
            val = grid.values[r, c]
            if np.isnan(val):
                ax.text(c, r, "—", ha="center", va="center",
                        fontsize=7, color="#aaa")
                continue
            # normalised position in [0,1] on the colour scale
            norm_val = (val - vmin) / (vmax - vmin)
            txt_col  = "white" if norm_val < 0.45 else "#1a1a1a"
            ax.text(c, r, f"{val:.2f}",
                    ha="center", va="center",
                    fontsize=7, color=txt_col, fontweight="medium")

    # White cell dividers
    for x in np.arange(-0.5, n_ra, 1):
        ax.axvline(x, color="white", lw=1.0, zorder=2)
    for y in np.arange(-0.5, n_pr, 1):
        ax.axhline(y, color="white", lw=1.0, zorder=2)

    ax.set_xlim(-0.5, n_ra - 0.5)
    ax.set_ylim(-0.5, n_pr - 0.5)

    # X ticks on every panel
    ax.set_xticks(range(n_ra))
    ax.set_xticklabels(ra_labels, fontsize=7, rotation=45, ha="right")

    # Y ticks only on leftmost panel
    if col == 0:
        ax.set_yticks(range(n_pr))
        ax.set_yticklabels(pr_labels, fontsize=7)
        ax.set_ylabel(r"Prandtl  $\mathrm{Pr}$", fontsize=9, labelpad=4)
    else:
        ax.set_yticks([])

    # Model title — bold + underline for hero
    weight = "bold" if model == HERO else "regular"
    title  = label + "\n*" if model == HERO else label
    ax.set_title(title, fontsize=8, fontweight=weight, pad=4, loc="center")

    # Hero panel: accent border
    if model == HERO:
        for spine in ax.spines.values():
            spine.set_edgecolor("#1D9E75")
            spine.set_linewidth(1.8)

# Shared x-axis label (centred under all panels)
fig.text(0.5, -0.02, r"Rayleigh  $\mathrm{Ra}$",
         ha="center", fontsize=9)

# ── 5. Single shared colorbar on the right ───────────────────────────────────

cbar_ax = fig.add_axes([1.01, 0.15, 0.025, 0.72])
sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm.set_array([])
cbar = fig.colorbar(sm, cax=cbar_ax)
cbar.set_label("VRMSE", fontsize=9, labelpad=8)
cbar.ax.tick_params(labelsize=7)

cbar.ax.text(0.5,  1.04, "high\nerror", ha="center", va="bottom",
             fontsize=6.5, color="#555", transform=cbar.ax.transAxes)
cbar.ax.text(0.5, -0.04, "low\nerror",  ha="center", va="top",
             fontsize=6.5, color=TEAL_DARK[:3], transform=cbar.ax.transAxes)

# ── 6. Supertitle ─────────────────────────────────────────────────────────────

fig.suptitle(
    r"VRMSE across physical regimes  (Ra $\times$ Pr)",
    fontsize=10, y=1.04, x=0.5,
)

#plt.savefig("panel_D_regime_heatmap.pdf", bbox_inches="tight", dpi=300)
#plt.savefig("panel_D_regime_heatmap.png", bbox_inches="tight", dpi=300)
#print("Saved: panel_D_regime_heatmap.pdf / .png")
plt.show()
```

```python
import math
import numpy as np
from matplotlib import ticker
import matplotlib.lines as mlines

# ── Config ────────────────────────────────────────────────────────────────
steps_to_show = [2, 4, 8, 16, 32, 64]
gt_label      = "Ground-truth"
field         = "buoyancy"           # fix one field; loop over field_names if needed

all_models    = [
    'WaveletFlow_1scale_db2_v4.0',
    'WaveletFlow_3scales_db2_v4.0',
    'FM_data_space_v4.0',
    gt_label,
]
non_gt_models = [m for m in all_models if m != gt_label]
wl_min, wl_max = 1, 128

# ── Step 1: unique (Ra, Pr) rows — sorted for reproducibility ─────────────
traj_combos = (
    psd_df[["Rayleigh", "Prandtl"]]
    .drop_duplicates()
    .sort_values(["Rayleigh", "Prandtl"])
    .reset_index(drop=True)
)
n_traj  = len(traj_combos)
n_steps = len(steps_to_show)

# ── Step 2: GT baseline per (rayleigh, prandtl, field, band, wavelength) ──
gt_psd = (
    psd_df[
        (psd_df["model"] == gt_label) &
        (psd_df["field"] == field)
    ]
    .groupby(["Rayleigh", "Prandtl", "band", "wavelength"])["psd"]
    .agg(aggregate)
    .reset_index()
    .rename(columns={"psd": "psd_gt"})
)

# ── Step 3: filter & merge GT ─────────────────────────────────────────────
ratio_ds = psd_df[
    psd_df["model"].isin(non_gt_models) &
    psd_df["step"].isin(steps_to_show) &
    (psd_df["field"] == field) &
    (psd_df["band"] != "vhigh") &
    (psd_df["wavelength"] >= wl_min) &
    (psd_df["wavelength"] <= wl_max)
].copy()

ratio_ds = ratio_ds.merge(
    gt_psd, on=["Rayleigh", "Prandtl", "band", "wavelength"], how="left"
)
ratio_ds["psd_ratio"] = ratio_ds["psd"] / ratio_ds["psd_gt"]

# ── Step 4: build figure — rows = (Ra,Pr) combos, cols = steps ────────────
fig, axs = plt.subplots(
    n_traj, n_steps,
    figsize=(3.2 * n_steps, 2.6 * n_traj),
    sharey="row",
    sharex=True,
    squeeze=False,
)

for row_idx, (_, traj) in enumerate(traj_combos.iterrows()):
    ra, pr = traj["Rayleigh"], traj["Prandtl"]
    df_traj = ratio_ds[
        (ratio_ds["Rayleigh"] == ra) &
        (ratio_ds["Prandtl"] == pr)
    ]

    for col_idx, step in enumerate(steps_to_show):
        ax    = axs[row_idx, col_idx]
        df_plot = df_traj[df_traj["step"] == step]

        # ── lines per model ───────────────────────────────────────────────
        sns.lineplot(
            data=df_plot,
            x="wavelength",
            y="psd_ratio",
            hue="model",
            hue_order=non_gt_models,
            palette={m: model_palette[m] for m in non_gt_models},
            estimator=aggregate,
            errorbar=("sd", 1),
            linewidth=1.5,
            ax=ax,
        )

        # ── reference & shading ───────────────────────────────────────────
        ax.axhline(1.0, color="black", linestyle=(0, (4, 2)), linewidth=1.2)
        ax.axhspan(0, 1.0, alpha=0.04, color="red", zorder=0)

        # ── log x axis & ticks ────────────────────────────────────────────
        ax.set_xscale("log", base=2)
        wl_vals = df_plot["wavelength"].replace([np.inf], np.nan).dropna()
        if len(wl_vals):
            exp_min = math.floor(math.log2(wl_vals.min()))
            exp_max = math.ceil(math.log2(wl_vals.max()))
            ticks   = [2**e for e in range(exp_min, exp_max + 1)]
            ax.set_xticks(ticks)
            ax.xaxis.set_major_formatter(
                ticker.FuncFormatter(
                    lambda v, _: rf"$2^{{{int(round(math.log2(v)))}}}$" if v > 0 else ""
                )
            )
        ax.xaxis.set_minor_locator(ticker.NullLocator())
        ax.invert_xaxis()

        # ── titles / labels ───────────────────────────────────────────────
        if row_idx == 0:
            ax.set_title(f"Step {step}", fontsize=10)
        if col_idx == 0:
            ax.set_ylabel(
                f"Ra={ra:.0e}, Pr={pr}\n" + r"$R(\lambda)=\hat{S}/S_{GT}$",
                fontsize=8,
            )
        else:
            ax.set_ylabel("")
        ax.set_xlabel("Wavelength" if row_idx == n_traj - 1 else "")

        # ── legend: only top-right cell ───────────────────────────────────
        if row_idx == 0 and col_idx == n_steps - 1:
            handles, labels = ax.get_legend_handles_labels()
            ref_handle = mlines.Line2D(
                [], [], color="black", linestyle=(0, (4, 2)),
                linewidth=1.2, label="GT reference (R=1)"
            )
            ax.legend(
                handles + [ref_handle],
                labels  + ["GT reference (R=1)"],
                fontsize=7, framealpha=0.7, loc="best",
            )
        else:
            leg = ax.get_legend()
            if leg:
                leg.remove()

        ax.grid(alpha=0.3)
        ax.set_ylim(bottom=0)

fig.suptitle(
    rf"Spectral ratio $R(\lambda)$ — field: {field} — closer to 1 is better",
    fontsize=11, y=1.01,
)
fig.tight_layout(pad=0.4)
plt.show()
```
