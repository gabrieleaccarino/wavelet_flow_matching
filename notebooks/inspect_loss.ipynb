```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
loss = pd.read_csv("/Users/gabrieleaccarino/Workspace/waveletflux/notebooks/metrics_am_1scale.csv", index_col=None)
```

```python
val_loss   = loss[~loss['val_loss'].isna() == True]['val_loss'].values
train_loss = loss['train_loss_epoch'].dropna().values
```

```python
plt.figure(figsize=(10, 5))
plt.plot(np.arange(200), train_loss, label='Train Loss')
plt.plot(np.arange(200), val_loss, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training and Validation Loss over Epochs')
plt.legend()
plt.grid()
plt.show()
```
